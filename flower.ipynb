```python
# ============================================
# 1. CÀI ĐẶT THƯ VIỆN
# ============================================
!pip install tensorflow matplotlib numpy scikit-learn seaborn -q

import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np
import os
import zipfile
import random
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout, BatchNormalization
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
from google.colab import drive
from pathlib import Path # Import Path

# ============================================
# 2. KẾT NỐI GOOGLE DRIVE (để lưu model sau khi train)
# ============================================
drive.mount('/content/drive')

# ============================================
# 3. TẢI DATASET (5 loài hoa)
# ============================
print("="*50)
print("ĐANG TẢI DATASET...")
print("="*50)

# Dataset chứa 5 loại hoa: daisy, dandelion, roses, sunflowers, tulips
# Nguồn: TensorFlow Datasets [citation:1][citation:6][citation:10]
dataset_url = "https://storage.googleapis.com/download.tensorflow.org/example_images/flower_photos.tgz"

# Tải và giải nén
# tf.keras.utils.get_file returns the path to the directory containing the extracted files (e.g., /root/.keras/datasets/flower_photos_extracted)
extracted_parent_dir = tf.keras.utils.get_file(
    fname='flower_photos.tgz',
    origin=dataset_url,
    extract=True
)

# Lấy đường dẫn thư mục chứa ảnh
# The actual dataset is in a subdirectory named 'flower_photos' inside the extracted_parent_dir
data_dir = Path(extracted_parent_dir) / 'flower_photos' # Correctly point to the 'flower_photos' subdirectory

# Kiểm tra số lượng ảnh
image_count = len(list(data_dir.glob('*/*.jpg')))
print(f"✅ Đã tải xong! Tổng số ảnh: {image_count}")

# Xem các lớp hoa
class_names = sorted([d for d in os.listdir(data_dir) if os.path.isdir(os.path.join(data_dir, d))])
print(f"📁 Các loại hoa: {class_names}")
print(f"📊 Số lượng ảnh mỗi loại:")
for cls in class_names:
    count = len(list(os.path.join(data_dir, cls, f) for f in os.listdir(os.path.join(data_dir, cls)) if f.endswith('.jpg')))
    print(f"   - {cls}: {count} ảnh")

# ============================================
# 4. HIỂN THỊ MỘT SỐ ẢNH MẪU
# ============================
print("\n" + "="*50)
print("HIỂN THỊ ẢNH MẪU")
print("="*50)

plt.figure(figsize=(12, 8))
for i, class_name in enumerate(class_names):
    class_dir = os.path.join(data_dir, class_name)
    images = [f for f in os.listdir(class_dir) if f.endswith('.jpg')]
    if images:
        img_path = os.path.join(class_dir, images[0])
        img = plt.imread(img_path)
        plt.subplot(2, 3, i+1)
        plt.imshow(img)
        plt.title(class_name)
        plt.axis('off')
plt.tight_layout()
plt.show()

# ============================================
# 5. TIỀN XỬ LÝ DỮ LIỆU & DATA AUGMENTATION
# ============================================
print("\n" + "="*50)
print("TIỀN XỬ LÝ DỮ LIỆU")
print("="*50)

IMG_HEIGHT = 150
IMG_WIDTH = 150
BATCH_SIZE = 32
VALIDATION_SPLIT = 0.2  # 20% cho validation

# Data augmentation cho training (giúp model tổng quát hóa tốt hơn)
train_datagen = ImageDataGenerator(
    rescale=1./255,           # Chuẩn hóa pixel về [0,1]
    rotation_range=40,        # Xoay ảnh ngẫu nhiên
    width_shift_range=0.2,    # Dịch ngang
    height_shift_range=0.2,   # Dịch dọc
    shear_range=0.2,          # Biến dạng
    zoom_range=0.2,           # Phóng to/thu nhỏ
    horizontal_flip=True,     # Lật ngang
    validation_split=VALIDATION_SPLIT
)

# Chỉ rescale cho validation (không augmentation)
val_datagen = ImageDataGenerator(
    rescale=1./255,
    validation_split=VALIDATION_SPLIT
)

# Tạo generator cho training set
train_generator = train_datagen.flow_from_directory(
    data_dir,
    target_size=(IMG_HEIGHT, IMG_WIDTH),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    subset='training',
    shuffle=True
)

# Tạo generator cho validation set
validation_generator = val_datagen.flow_from_directory(
    data_dir,
    target_size=(IMG_HEIGHT, IMG_WIDTH),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    subset='validation',
    shuffle=False
)

print(f"\n✅ Training set: {train_generator.samples} ảnh")
print(f"✅ Validation set: {validation_generator.samples} ảnh")
print(f"📋 Class indices: {train_generator.class_indices}")

# ============================================
# 6. XÂY DỰNG MÔ HÌNH CNN
# ============================
print("\n" + "="*50)
print("XÂY DỰNG MÔ HÌNH CNN")
print("="*50)

model = Sequential([
    # Block 1
    Conv2D(32, (3, 3), activation='relu', padding='same', input_shape=(IMG_HEIGHT, IMG_WIDTH, 3)),
    BatchNormalization(),
    Conv2D(32, (3, 3), activation='relu', padding='same'),
    BatchNormalization(),
    MaxPooling2D(2, 2),

    # Block 2
    Conv2D(64, (3, 3), activation='relu', padding='same'),
    BatchNormalization(),
    Conv2D(64, (3, 3), activation='relu', padding='same'),
    BatchNormalization(),
    MaxPooling2D(2, 2),

    # Block 3
    Conv2D(128, (3, 3), activation='relu', padding='same'),
    BatchNormalization(),
    Conv2D(128, (3, 3), activation='relu', padding='same'),
    BatchNormalization(),
    MaxPooling2D(2, 2),

    # Block 4
    Conv2D(256, (3, 3), activation='relu', padding='same'),
    BatchNormalization(),
    MaxPooling2D(2, 2),

    # Fully Connected Layers
    Flatten(),
    Dropout(0.5),
    Dense(512, activation='relu'),
    Dropout(0.3),
    Dense(256, activation='relu'),
    Dense(5, activation='softmax')  # 5 classes
])

# Compile model
model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()

# ============================================
# 7. CALLBACKS
# ============================
# Lưu model tốt nhất vào Google Drive
checkpoint = ModelCheckpoint(
    '/content/drive/MyDrive/best_flower_model.h5',
    monitor='val_accuracy',
    save_best_only=True,
    mode='max',
    verbose=1
)

# Dừng sớm nếu validation loss không giảm
early_stop = EarlyStopping(
    monitor='val_loss',
    patience=10,
    restore_best_weights=True,
    verbose=1
)

# Giảm learning rate khi loss plateaus
reduce_lr = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.2,
    patience=5,
    min_lr=0.00001,
    verbose=1
)

# ============================================
# 8. HUẤN LUYỆN MÔ HÌNH
# ============================
print("\n" + "="*50)
print("BẮT ĐẦU HUẤN LUYỆN")
print("="*50)

EPOCHS = 50

history = model.fit(
    train_generator,
    steps_per_epoch=train_generator.samples // BATCH_SIZE,
    epochs=EPOCHS,
    validation_data=validation_generator,
    validation_steps=validation_generator.samples // BATCH_SIZE,
    callbacks=[early_stop, checkpoint, reduce_lr],
    verbose=1
)

# ============================================
# 9. VẼ BIỂU ĐỒ KẾT QUẢ
# ============================
print("\n" + "="*50)
print("BIỂU ĐỒ KẾT QUẢ")
print("="*50)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Accuracy
ax1.plot(history.history['accuracy'], 'b-', label='Train Accuracy')
ax1.plot(history.history['val_accuracy'], 'r-', label='Validation Accuracy')
ax1.set_title('Model Accuracy', fontsize=14)
ax1.set_xlabel('Epoch')
ax1.set_ylabel('Accuracy')
ax1.legend()
ax1.grid(True)

# Loss
ax2.plot(history.history['loss'], 'b-', label='Train Loss')
ax2.plot(history.history['val_loss'], 'r-', label='Validation Loss')
ax2.set_title('Model Loss', fontsize=14)
ax2.set_xlabel('Epoch')
ax2.set_ylabel('Loss')
ax2.legend()
ax2.grid(True)

plt.tight_layout()
plt.show()

# ============================================
# 10. ĐÁNH GIÁ MÔ HÌNH
# ==============================
print("\n" + "="*50)
print("ĐÁNH GIÁ MÔ HÌNH")
print("="*50)

# Evaluate trên validation set
val_loss, val_acc = model.evaluate(validation_generator, verbose=0)
print(f"📊 Validation Accuracy: {val_acc:.4f}")
print(f"📊 Validation Loss: {val_loss:.4f}")

# Dự đoán toàn bộ validation set
validation_generator.reset()
Y_pred = model.predict(validation_generator, steps=validation_generator.samples // BATCH_SIZE + 1)
y_pred = np.argmax(Y_pred, axis=1)
y_true = validation_generator.classes[:len(y_pred)]

# Classification Report
class_names_list = list(train_generator.class_indices.keys())
print("\n" + "="*50)
print("CLASSIFICATION REPORT")
print("="*50)
print(classification_report(y_true, y_pred, target_names=class_names_list))

# Confusion Matrix
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=class_names_list,
            yticklabels=class_names_list)
plt.title('Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()

# ============================================
# 11. LƯU MODEL
# ============================
print("\n" + "="*50)
print("LƯU MODEL")
print("="*50)

# Lưu model cuối cùng
model.save('/content/drive/MyDrive/final_flower_model.h5')
print("✅ Model đã được lưu tại: /content/drive/MyDrive/final_flower_model.h5")

# Lưu class labels
import json
class_labels = train_generator.class_indices
with open('/content/drive/MyDrive/class_labels.json', 'w') as f:
    json.dump(class_labels, f)
print("✅ Class labels đã được lưu tại: /content/drive/MyDrive/class_labels.json")

# ============================================
# 12. HÀM DỰ ĐOÁN CHO ẢNH MỚI
# ============================
print("\n" + "="*50)
print("HÀM DỰ ĐOÁN")
print("="*50)

def predict_flower(image_path, model_path=None, class_labels_path=None):
    """
    Dự đoán loại hoa từ ảnh

    Args:
        image_path: Đường dẫn đến ảnh (có thể upload lên Colab)
        model_path: Đường dẫn đến model (mặc định dùng model vừa train)
        class_labels_path: Đường dẫn đến file class labels

    Returns:
        predicted_flower: Tên loài hoa dự đoán
        confidence: Độ tin cậy
        all_probabilities: Xác suất của tất cả các lớp
    """
    from tensorflow.keras.preprocessing import image

    # Load model nếu cần
    if model_path is None:
        model_path = '/content/drive/MyDrive/final_flower_model.h5'

    if class_labels_path is None:
        class_labels_path = '/content/drive/MyDrive/class_labels.json'

    # Load class labels
    with open(class_labels_path, 'r') as f:
        class_labels = json.load(f)

    # Đảo ngược dictionary để lấy tên từ index
    class_names = {v: k for k, v in class_labels.items()}

    # Load và xử lý ảnh
    img = image.load_img(image_path, target_size=(IMG_HEIGHT, IMG_WIDTH))
    img_array = image.img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0)
    img_array = img_array / 255.0

    # Dự đoán
    predictions = model.predict(img_array, verbose=0)
    predicted_class = np.argmax(predictions[0])
    confidence = np.max(predictions[0])
    predicted_flower = class_names[predicted_class]

    # Hiển thị kết quả
    print(f"\n🔮 Kết quả dự đoán:")
    print(f"   🌸 Đây là hoa: {predicted_flower}")
    print(f"   📈 Độ tin cậy: {confidence:.2%}")

    # Hiển thị tất cả xác suất
    print(f"\n📊 Xác suất cho từng loại:")
    for i, prob in enumerate(predictions[0]):
        print(f"   {class_names[i]}: {prob:.2%}")

    return predicted_flower, confidence, predictions[0]

# ============================================
# 13. TEST THỬ VỚI ẢNH TỪ VALIDATION SET
# ============================
print("\n" + "="*50)
print("TEST THỬ VỚI ẢNH MẪU")
print("="*50)

# Lấy một batch từ validation generator
test_images, test_labels = next(validation_generator)

# Chọn ngẫu nhiên 5 ảnh để test
num_samples = 5
indices = random.sample(range(len(test_images)), num_samples)

class_names_rev = {v: k for k, v in train_generator.class_indices.items()}

fig, axes = plt.subplots(1, num_samples, figsize=(15, 4))

for idx, i in enumerate(indices):
    # Dự đoán
    img = np.expand_dims(test_images[i], axis=0)
    pred = model.predict(img, verbose=0)
    pred_class = np.argmax(pred)
    pred_label = class_names_rev[pred_class]
    confidence = np.max(pred)

    # True label
    true_label = class_names_rev[np.argmax(test_labels[i])]

    # Màu chữ: xanh nếu đúng, đỏ nếu sai
    color = 'green' if pred_label == true_label else 'red'

    # Hiển thị
    axes[idx].imshow(test_images[i])
    axes[idx].set_title(f"True: {true_label}\nPred: {pred_label}\nConf: {confidence:.2%}", color=color)
    axes[idx].axis('off')

plt.tight_layout()
plt.show()

```

```python
!pip install gradio -q

import gradio as gr
import numpy as np
import json
import requests
from io import BytesIO
from PIL import Image
import tensorflow as tf
import matplotlib
matplotlib.use('Agg')

# ============================================
# CẤU HÌNH (phải khớp với code training)
# ============================================
IMG_HEIGHT = 150
IMG_WIDTH  = 150
MODEL_PATH  = '/content/drive/MyDrive/final_flower_model.h5'
LABELS_PATH = '/content/drive/MyDrive/class_labels.json'

FLOWER_META = {
    'daisy':       {'emoji': '🌼', 'vi': 'Hoa Cúc'},
    'dandelion':   {'emoji': '🌻', 'vi': 'Bồ Công Anh'},
    'roses':       {'emoji': '🌹', 'vi': 'Hoa Hồng'},
    'sunflowers':  {'emoji': '🌻', 'vi': 'Hướng Dương'},
    'tulips':      {'emoji': '🌷', 'vi': 'Hoa Tulip'},
}

# ============================================
# LOAD MODEL & CLASS LABELS
# ============================================
print("⏳ Đang tải model...")
loaded_model = tf.keras.models.load_model(MODEL_PATH)

with open(LABELS_PATH, 'r') as f:
    class_indices = json.load(f)          # {"daisy": 0, "dandelion": 1, ...}

class_names = {v: k for k, v in class_indices.items()}   # {0: "daisy", ...}
print(f"✅ Model đã tải! Nhận diện {len(class_names)} loài hoa.")

# ============================================
# HÀM DỰ ĐOÁN DÙNG CHUNG
# ============================================
def run_prediction(pil_img):
    """Nhận PIL Image, trả về dict kết quả."""
    img = pil_img.convert('RGB').resize((IMG_WIDTH, IMG_HEIGHT))
    arr = np.expand_dims(np.array(img) / 255.0, axis=0)
    probs = loaded_model.predict(arr, verbose=0)[0]

    top_idx   = int(np.argmax(probs))
    top_name  = class_names[top_idx]
    top_prob  = float(probs[top_idx])
    meta      = FLOWER_META.get(top_name, {'emoji': '🌸', 'vi': top_name})

    # Dict cho thanh xác suất của Gradio
    label_probs = {
        f"{FLOWER_META.get(class_names[i], {}).get('emoji','🌸')} "
        f"{FLOWER_META.get(class_names[i], {}).get('vi', class_names[i])}": float(probs[i])
        for i in range(len(probs))
    }

    summary = (
        f"## {meta['emoji']} {meta['vi'].upper()}\n\n"
        f"**Độ tin cậy:** {top_prob:.1%}\n\n"
        f"**Tên khoa học:** *{top_name}*"
    )

    return summary, label_probs


# ============================================
# HÀM GỌI TỪ GRADIO (UPLOAD)
# ============================================
def predict_upload(image):
    """Nhận diện từ ảnh upload."""
    if image is None:
        return "⚠️ Vui lòng tải lên một ảnh!", {}
    try:
        pil_img = Image.fromarray(image)
        return run_prediction(pil_img)
    except Exception as e:
        return f"❌ Lỗi: {e}", {}


# ============================================
# HÀM GỌI TỪ GRADIO (URL)
# ============================================
def predict_url(url):
    """Nhận diện từ URL hình ảnh."""
    url = url.strip()
    if not url:
        return "⚠️ Vui lòng nhập URL hình ảnh!", {}, None

    try:
        headers = {'User-Agent': 'Mozilla/5.0'}
        resp = requests.get(url, headers=headers, timeout=10)

        content_type = resp.headers.get('Content-Type', '')
        if 'image' not in content_type:
            return (
                "❌ URL này không phải ảnh trực tiếp.\n\n"
                "💡 Hãy dùng URL kết thúc bằng .jpg / .png",
                {}, None
            )

        pil_img = Image.open(BytesIO(resp.content)).convert('RGB')
        summary, label_probs = run_prediction(pil_img)
        return summary, label_probs, np.array(pil_img)

    except Exception as e:
        return f"❌ Lỗi tải ảnh: {e}", {}, None


# ============================================
# GIAO DIỆN GRADIO
# ============================================
CSS = """
#title  { text-align: center; }
#result { font-size: 1.1rem; }
"""

with gr.Blocks(theme=gr.themes.Soft(), css=CSS, title="🌸 Nhận Diện Hoa") as demo:

    # ---- Header ----
    gr.Markdown(
        "# 🌸 Nhận Diện 5 Loài Hoa\n"
        "Upload ảnh hoặc dán URL để nhận diện loài hoa bằng mô hình CNN",
        elem_id="title"
    )

    gr.Markdown(
        "> **5 loài:** 🌼 Hoa Cúc &nbsp;|&nbsp; 🌻 Bồ Công Anh &nbsp;|&nbsp; "
        "🌹 Hoa Hồng &nbsp;|&nbsp; 🌻 Hướng Dương &nbsp;|&nbsp; 🌷 Tulip"
    )

    # ---- Tabs ----
    with gr.Tabs():

        # === TAB 1: UPLOAD ===
        with gr.Tab("📁 Tải Ảnh Lên"):
            with gr.Row():
                with gr.Column(scale=1):
                    inp_image = gr.Image(
                        label="Chọn ảnh hoa",
                        type="numpy",
                        sources=["upload", "clipboard"]
                    )
                    btn_upload = gr.Button(
                        "🔍 Nhận Diện", variant="primary", size="lg"
                    )

                with gr.Column(scale=1):
                    out_md_upload = gr.Markdown(
                        label="Kết quả", elem_id="result",
                        value="*Kết quả sẽ hiện ở đây...*"
                    )
                    out_bar_upload = gr.Label(
                        label="Xác suất từng loài", num_top_classes=5
                    )

            btn_upload.click(
                fn=predict_upload,
                inputs=inp_image,
                outputs=[out_md_upload, out_bar_upload]
            )

            # Ví dụ nhanh
            gr.Examples(
                examples=[
                    ["https://images.unsplash.com/photo-1490750967868-88aa4486c946?w=400"],
                ],
                inputs=inp_image,
                label="Ảnh mẫu"
            )

        # === TAB 2: URL ===
        with gr.Tab("🔗 Dán URL"):
            with gr.Row():
                with gr.Column(scale=1):
                    inp_url = gr.Textbox(
                        label="URL hình ảnh hoa",
                        placeholder="https://images.unsplash.com/photo-...?w=400",
                        lines=2
                    )
                    btn_url = gr.Button(
                        "🔍 Nhận Diện", variant="primary", size="lg"
                    )
                    out_img_url = gr.Image(
                        label="Ảnh từ URL", type="numpy", interactive=False
                    )

                with gr.Column(scale=1):
                    out_md_url = gr.Markdown(
                        label="Kết quả", elem_id="result",
                        value="*Kết quả sẽ hiện ở đây...*"
                    )
                    out_bar_url = gr.Label(
                        label="Xác suất từng loài", num_top_classes=5
                    )

            btn_url.click(
                fn=predict_url,
                inputs=inp_url,
                outputs=[out_md_url, out_bar_url, out_img_url]
            )

            # URL mẫu sẵn
            gr.Examples(
                examples=[
                    ["https://images.unsplash.com/photo-1582794314132-3d82e8549e91?w=400"],
                    ["https://images.unsplash.com/photo-1490750967868-88aa4486c946?w=400"],
                    ["https://images.unsplash.com/photo-1597848856608-b86b48235b6c?w=400"],
                    ["https://images.unsplash.com/photo-1583390039828-c31f4c02eb82?w=400"],
                ],
                inputs=inp_url,
                label="URL mẫu (click để dùng ngay)"
            )

    # ---- Footer ----
    gr.Markdown(
        "---\n"
        "💡 **Mẹo tìm URL ảnh:** Vào Unsplash → chọn ảnh → chuột phải → *Copy image address*"
    )


# ============================================
# CHẠY ỨNG DỤNG
# ============================================
demo.launch(
    share=True,        # Tạo link công khai chia sẻ được
    debug=False,
    quiet=True
)
```
